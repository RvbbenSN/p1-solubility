### 4. Summary Statistics for Measured Solubility (Target variable $y$)
The primary regression target is `measured log solubility in mols per litre`.

```python
target_col = "measured log solubility in mols per litre"
df[target_col].describe()
```

### 5. Extreme Values in Chemical Space
- **Most soluble molecule in water:** (highest log S)
- **Most insoluble molecule in water:** (lowest log S)

```python
# Most soluble compound
most_soluble = df.loc[df[target_col].idxmax()]
print("=== MOST SOLUBLE COMPOUND ===")
print(f"Compound ID: {most_soluble['Compound ID']}")
print(f"SMILES:      {most_soluble['smiles']}")
print(f"Measured logS: {most_soluble[target_col]} mol/L
")

# Most insoluble compound
least_soluble = df.loc[df[target_col].idxmin()]
print("=== MOST INSOLUBLE COMPOUND ===")
print(f"Compound ID: {least_soluble['Compound ID']}")
print(f"SMILES:      {least_soluble['smiles']}")
print(f"Measured logS: {least_soluble[target_col]} mol/L")
```
